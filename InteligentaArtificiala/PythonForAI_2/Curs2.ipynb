## Collection comprehension

Pornind de la o colecție (cel mai frecvent caz: de la o listă) se poate crea o altă colecție, folosind *collection comprehension* - în esență o ciclare:

````python
lista_numere = [1, 3, 6, 21, 22, 32, 33]
lista_patrate = [i * i for i in lista_numere]
print(lista_patrate)
````

Opțional, la fiecare pas al iterării se poate lua în considerare o condiție `if` inline:

````python
lista_patrate_doar_numere_pare = [i * i for i in lista_numere if i % 2 == 0]
print(lista_patrate_doar_numere_pare)
````

... sau se folosește și `else` pe lângă `if`:

````python
lista_patrate_sau_cuburi = [i**2 if i % 2 == 0 else i**3 for i in lista_numere]
print(lista_patrate_sau_cuburi)
````

Exercițiu: dacă o listă are ca elemente alte liste de numere, cum se poate determina lista 'flattened'? De exemplu, pentru lista `a1 = [[1, 2], [3, 4, 5], [10]]` se dorește rezultatul `a2 = [1, 2, 3, 4, 5, 10]`.

````python
a1 = [[1, 2], [3, 4, 5], [10]]
a2 = [item for sublist in a1 for item in sublist]
print(a2)
````

Comprehension se poate folosi și peste alte tipuri de colecții: de exemplu, putem pleca de la o listă și producem un dicționar:

````python
lista_numere = [1, 3, 6, 21, 22, 32, 33]
dictionar_numere_si_patrate = {i: i**2 for i in lista_numere}
print(dictionar_numere_si_patrate)
````

... sau liste de tupluri:

````python
# produs cartezian
colours = ["red", "green", "yellow", "blue"]
things = ["house", "car", "tree"]
produs_cartezian = [(colour, thing) for colour in colours for thing in things]
print(produs_cartezian)
assert len(produs_cartezian) == len(colours) * len(things)
````

Mai jos sunt câteva exemple de utilizare de comprehension peste colecții.

````python
# Conversie de temperaturi din Celsius in Fahrenheit: valoarea in Fahrenheit se obtine cu formula: 1.8 * gradeCelsius + 32
grade_celsius = [-20, -10, 0, 5, 23, 35]
grade_fahrenheit = [1.8 * gc + 32 for gc in grade_celsius]
print(grade_fahrenheit)
````

````python
# Suma patratelor numerelor de la 1 la 20
print(sum([x**2 for x in range(1, 21)]))
````

## Funcții

Funcțiile sunt de trei feluri:
* Funcții deja definite în limbajul Python, cum ar fi `len()`, `print()`
* Funcții definite de utilizator
* Lambda funcții

O funcție se definește folosind cuvântul cheie `def`. Blocul de instrucțiuni ce definește corpul funcției este indentat. Urmează câteva exemple de funcții definite de utilizator, cu comentarii:

````python
def hello_with_name(nume):
    """
    Functia preia un argument si afiseaza mesajul: Salutare urmat de valoarea argumentului.
    Functia returneaza argumentul cu litere mari.
    :param nume: numele care se cere afisat
    :return: sirul din :param nume: cu litere mari
    """
    print("Salutare " + nume)
    return nume.upper()


nume = "Natalia"
nume_litere_mari = hello_with_name(nume)
print(nume_litere_mari)
help(hello_with_name)
print(hello_with_name.__doc__)
````

Pot exista parametri cu valori implicite, precizați la finalul listei de parametri formali:

#### Capcană: valori implicite mutabile

Valoarea implicită a unui parametru se calculează o singură dată, atunci când se execută instrucțiunea `def`, și nu la fiecare apel. Dacă această valoare este un obiect mutabil (listă, dicționar, mulțime), toate apelurile care nu precizează parametrul folosesc același obiect. Dacă la un apel se modifică valoarea implicită, ea rămâne vizibilă și în apelurile următoare:


````python
# varianta gresita: lista implicita este creata o singura data, la definirea functiei
def adauga_nota(nota, note=[]):
    note.append(nota)
    return note


print(adauga_nota(10))
print(adauga_nota(7))  # ne-am astepta la [7]
print(adauga_nota(9))

# valoarea implicita este pastrata in obiectul functie si se vede cum s-a modificat
print(adauga_nota.__defaults__)
````

Lista implicită este păstrată în obiectul funcție (atributul `__defaults__`) și este aceeași la fiecare apel. Soluția uzuală este ca valoarea implicită să fie `None`, iar obiectul mutabil să fie creat în corpul funcției:


````python
# varianta corecta: None ca valoare implicita, lista se creeaza la fiecare apel
def adauga_nota(nota, note=None):
    if note is None:
        note = []
    note.append(nota)
    return note


print(adauga_nota(10))
print(adauga_nota(7))
print(adauga_nota(9, [8, 6]))
````

De reținut:
* ca valori implicite se folosesc doar obiecte imuabile: `None`, numere, șiruri de caractere, `True`/`False`, tupluri;
* se evită `[]`, `{}`, `set()` și, în general, valorile obținute printr-un apel de funcție, care s-ar calcula o singură dată (de exemplu `datetime.now()`);
* când e nevoie de o listă sau de un dicționar implicit, se scrie `parametru=None`, iar obiectul se creează în funcție.

Detalii: [Default Argument Values](https://docs.python.org/3/tutorial/controlflow.html#default-argument-values) în tutorialul Python.


### Funcții cu număr arbitrar de argumente

Putem avea un parametru care să permită un număr variabil de valori trimise la apel; acest tip de parametru se scrie cu `*` urmat de numele parametrului formal (de exemplu: `*args`).

````python
# Functie cu numar arbitrar de argumente
def greet(*names, msg="buna dimineata!"):
    for name in names:
        print("Servus", name + ", " + msg)


greet("Dan", "John", "Mary")
greet("Dan", "John", "Mary", msg="ce mai faci?")
````

Se pot defini funcții care să manipuleze un număr variabil de parametri dați la apel sub formă de `nume_parametru=valoare_parametru`; denumirea tradițională este `kwargs` (keyword arguments), numele parametrului se prefixează cu `**`:

````python
def demo_kwargs(**kwargs):
    print(kwargs)


demo_kwargs(fruits="apples", quantity="3", measurement_unit="kg")
````

Iterarea peste perechile din `kwargs` se face ținând cont că acesta este un dicționar:

````python
def demo_kwargs_iter(**kwargs):
    for key, value in kwargs.items():
        print(key, value)


demo_kwargs_iter(fruits="apples", quantity="3", measurement_unit="kg")
````

Același efect se obține prin despachetarea de dicționare, folosind `**`:

````python
dictionar_argumente = {"fruits": "apples", "quantity": "3", "measurement_unit": "kg"}
demo_kwargs(**dictionar_argumente)
````

Ordonarea parametrilor declarați într-o funcție este:
1. parametri formali preluați prin poziție
1. `*args`
1. parametri cu valori implicite
1. `**kwargs`

```python
def example2(arg_1, arg_2, *args, kw_1="shark", kw_2="blobfish", **kwargs):
```

### Lambda funcții

Se pot defini funcții anonime (sau: lambda funcții), conținând o expresie, pentru care nu se consideră necesară definirea unor funcții separate. O lambda funcție poate să preia oricâte argumente și calculează o expresie pe baza lor. Lambda funcțiile pot accesa atât parametrii trimiși, cât și variabilele globale. Se va omite cuvântul `return`, expresia calculată este cea care se returnează automat. 

````python
suma = lambda x, y: x + y
print(suma(3, 4))
````

````python
# lambda functie pentru filtrare via functia filter
lista_30 = list(range(30))
lista_filtrata = list(filter(lambda x: x % 3 == 0, lista_30))
print(lista_filtrata)
````

````python
# lambda functie pentru sortare:
sorted([-1, -2, -3, 2, 3, 4, -5, 6, 7, 8, 9], key=lambda x: x**2)
````

### Funcții callback

Numele unei funcții reprezintă un obiect funcție:

````python
def sum_2(x, y):
    return x + y


def dif_2(x, y):
    return x - y


# reprezentarea prin functia repr este implementata implicit prin adresa de memorie
print(dif_2)
````

Putem folosi acest mecanism pentru a trimite funcții ca parametri într-o altă funcție:

````python
def complex_operation(x, y, to_be_called):
    return to_be_called(x, y)


print(complex_operation(2, 3, sum_2))
print(complex_operation(2, 3, dif_2))
````

## Module

Modulele sunt fișiere Python cu extensia .py, în care se găsesc implementări de funcții, clase, declarații de variabile. Importarea unui modul se face cu instrucțiunea `import`. 

Exemplu: creăm un modul - fișierul Python mySmartModule.py - care conține o funcție ce calculează suma elementelor dintr-o listă:


```python
# fisierul mySmartModule.py
def my_sum(lista):
    result = 0
    for item in lista:
        result += item
    return result
```

Utilizarea se face cu:
```python
import mySmartModule

lista = [1, 2, 3]

suma = mySmartModule.my_sum(lista)
print(suma)
```

Se poate ca modulul să fie importat cu un nume mai scurt, sub forma:
```python
import mySmartModule as msm
```
și în acest caz apelul se face cu:
```python
suma = msm.my_sum(lista)
```

Putem afla ce pune la dispoziție un modul:
```python
>>> dir(msm)
['__builtins__', '__cached__', '__doc__', '__file__', '__loader__', '__name__', '__package__', '__spec__', 'my_sum']
```
Elementele aflate între dublu underscore ('dunders') sunt adăugate automat de Python. 

Dacă se dorește ca tot ceea ce e definit într-un modul să fie disponibil fără a mai face prefixare cu `nume_modul.nume_entitate`, atunci se poate proceda astfel:
```python
from mySmartModule import *

print(my_sum([1, 2, 3]))
```
Se recomandă însă să se importe strict acele entități (funcții, tipuri) din modul care sunt utilizate; în felul acesta se evită suprascrierea prin import a altor entități deja importate:
```python
from mySmartModule import my_sum

print(my_sum([1, 2, 3]))
```

Ordinea de căutare a modulelor este:
1. directorul curent
1. dacă nu se găsește modulul cerut, se caută în variabila de mediu `PYTHONPATH`, dacă e definită
1. dacă nu se găsește modulul cerut, se caută în calea implicită.

Calea de căutare implicită se găsește în variabila `path` din modulul sistem `sys`:

````python
import sys

print(sys.path)
````

Dacă se dorește ca un modul scris de utilizator într-un director ce nu se găsește în lista de mai sus să fie accesibil pentru import, atunci calea către director trebuie adăugată la colecția `sys.path`:

````python
sys.path.append("./my_modules/")
from mySmartModule import my_sum

print(my_sum([1, 2, 3]))
````

Un modul se poate folosi în două feluri: 
1. pentru a pune la dispoziție diferite implementări de funcții sau de clase, sau variabile setate la anumite valori (de exemplu `math.pi`):
```python
import math
print(math.pi)
```
2. se poate lansa de sine stătător, scriind în linia de comandă: `python mySmartModule.py`. Pentru acest caz, dacă se vrea să se execute o anumită secvență de cod, atunci se va scrie în modulul Python `mySmartModule.py`:
```python
if __name__ == '__main__':
    #cod care se executa la lansarea directa a script-ului
```

Codul din secțiunea `if` scrisă ca mai sus nu se va executa când modulul este importat.

Exemplu:
```python
def my_sum(lista):
    result = 0
    for item in lista:
        result += item
    return result

if __name__ == '__main__':
    print('Exemplu de utilizare')
    lista = list(range(100))
    print(my_sum(lista))
```

## Pachete Python

Un pachet este o colecție de module. Din punct de vedere fizic, un pachet este o structură de directoare în care se găsesc module și alte pachete. Un director devine pachet atunci când conține un fișier `__init__.py`, care poate fi gol. Din Python 3.3 sunt acceptate și pachete fără acest fișier (*namespace packages*, PEP 420), dar pentru pachetele proprii se recomandă păstrarea lui.

Plecăm de la structura de directoare și fișiere:
```
---myUtils\
 |------ mySmartModule.py
 |------ __init__.py
```

Pentru importul funcției `my_sum` din fișierul `mySmartModule.py` aflat în directorul `myUtils` - care se dorește a fi pachet - s-ar scrie astfel:
```python
from myUtils.mySmartModule import my_sum
print(my_sum([1, 2, 3]))
```
dar am prefera să putem scrie:
```python
from myUtils import my_sum
print(my_sum([1, 2, 30]))
```
adică să nu mai referim modulul (fișierul) `mySmartModule` din cadrul pachetului `myUtils`. Pentru asta vom adăuga în fișierul `__init__.py` din directorul `myUtils` linia:
```python
from .mySmartModule import my_sum 
```
unde caracterul `.` de dinaintea numelui de modul `mySmartModule` se referă la calea relativă. 

````python
from myUtils import my_sum

print(my_sum([1, 2, 10, 300]))
````

În fișierul `__init__.py` se pun de regulă importurile care definesc ce oferă pachetul și, eventual, constante. Operațiile costisitoare, cum ar fi încărcarea de date de pe disc, se evită aici, pentru că se execută la fiecare `import`; ele se fac la cerere (lazy loading), în funcțiile care au nevoie de date.

Pentru cazul în care se dorește crearea de pachete destinate comunității și publicarea pe PyPI, se poate urma tutorialul oficial [Packaging Python Projects](https://packaging.python.org/en/latest/tutorials/packaging-projects/). Acesta descrie configurarea proiectului prin fișierul `pyproject.toml`, construirea pachetului cu `python -m build` și publicarea cu `twine`, mai întâi pe serverul de test TestPyPI. Pentru cei care folosesc `uv`, același flux se reduce la comenzile `uv build` și `uv publish`: [Building and publishing a package](https://docs.astral.sh/uv/guides/package/).

Alte exemple de utilizare de pachete sunt:

````python
import re  # pachet pentru expresii regulate

my_string = "Am cumparat: mere, pere, prune... si caise"
tokens = re.split(r"\W+", my_string)
print(tokens)
````

````python
import pickle
from pathlib import Path

favourite_colour = {"lion": "yellow", "kitty": "red"}
with open("save.pkl", "wb") as f:
    pickle.dump(favourite_colour, f)
del favourite_colour

with open("save.pkl", "rb") as f:
    favourite_colour_restored = pickle.load(f)
print("dupa deserializare:", favourite_colour_restored)

Path("save.pkl").unlink()  # sterge fisierul, indiferent de sistemul de operare
````

**Atenție:** `pickle.load` execută cod în timpul încărcării. Un fișier pickle primit din altă sursă poate rula orice comandă pe calculatorul vostru. Se încarcă doar fișiere pickle create de voi sau de o sursă de încredere; pentru schimbul de date se folosesc formate care conțin doar date: CSV, JSON, `np.save` / `np.load`. Același risc există la `joblib.load` și `torch.load`, care folosesc intern pickle. Detalii: avertismentul din [documentația oficială](https://docs.python.org/3/library/pickle.html).

## Pachetul NumPy

NumPy (Numerical Python) este pachetul de bază pentru calcule științifice în Python. Asigură suport pentru lucrul cu vectori și matrice multidimensionale, funcții dedicate precum sortare, operații din algebra liniară, procesare de semnal, calcule statistice de bază, generare de numere aleatoare etc. NumPy stă la baza multor altor pachete. Datele pe care le procesează trebuie să încapă în memoria RAM. NumPy are la bază cod C compilat și optimizat. 

În destul de multe situații, datele sunt sau pot fi transformate în numere:
* o imagine în tonuri de gri poate fi văzută ca o matrice bidimensională de numere; fiecare număr reprezintă intensitatea pixelului (0 - negru, 255 - alb; frecvent se folosesc valorile scalate în intervalul [0, 1])
![Imagine MNIST scalată 0-1](./images/mnist_image.png)
* o imagine color poate fi văzută ca o matrice cu trei dimensiuni: 3 matrice bidimensionale "paralele", corespunzătoare canalelor red, green, blue; pentru fiecare canal de culoare valorile pot fi între 0 și 255:
![Imagine RGB descompusă în 3 canale](./images/channelsrgb.gif)
* un fișier audio este văzut ca unul/doi/$k$ vectori unidimensionali, corespunzători cazurilor: mono, stereo, $k$ canale. Valorile numerice în cazul unui fișier wav reprezintă deplasarea membranei microfonului, discretizată în timp și amplitudine:
![sunet](./images/sound.png)
* un text poate fi tradus în vectori numerici prin tehnici precum [Bag of words](https://en.wikipedia.org/wiki/Bag-of-words_model) sau [Word2vec](https://en.wikipedia.org/wiki/Word2vec).

Reprezentarea este mult mai eficientă decât pentru listele Python; codul scris cu NumPy apelează biblioteci compilate în cod nativ. Dacă se scrie cod vectorizat, eficiența rulării e și mai mare. 

Tipul cel mai comun din NumPy este `ndarray` - n-dimensional array. 

````python
# import de pachet; traditional se foloseste abrevierea np pentru numpy
import numpy as np

# crearea unui vector pornind de la o lista Python
x = np.array([1, 4, 2, 5, 3])

# tipul variabilei x; se observa ca e tip numpy
print(type(x))
# toate elementele din array sunt de acelasi tip
print(x.dtype)

# specificarea explicita a tipului de reprezentare a datelor in array
y = np.array([1, 2, 3], dtype=np.float16)
print(y.dtype)
````

### Indexarea cu liste de indici (fancy indexing)

În locul unui indice sau al unui domeniu `start:stop:pas` se poate da o listă (sau un tablou) de indici. Rezultatul conține elementele de pe pozițiile cerute, în ordinea din listă; un indice poate apărea de mai multe ori:

Pentru matrice, o listă de indici pe o axă selectează linii sau coloane întregi. Dacă se dau liste pe ambele axe, ele se împerechează element cu element: `matrice[[0, 2], [1, 3]]` înseamnă elementele `[0, 1]` și `[2, 3]`, nu blocul format din liniile 0, 2 și coloanele 1, 3. ### Diferente dpdv al memoriei intre slicing si indexare

Slicing-ul returnează un 'view' al tabloului, peste care se pot aplica modificări ale conținutului originar:

````python
print("Inainte:\n", tab)

linie = tab[1, :]  # slicing -> view: aceleasi date ca tab
linie[:] = -100
print("Dupa modificarea view-ului:\n", tab)  # tab s-a modificat
````

Operarea pe view înseamnă modificarea valorilor corespunzătoare din tabloul inițial, deoarece view și tabloul partajează memorie. Indexarea logică și cea cu liste de indici, însă, produc o copie; modificarea elementelor din copie nu afectează zona de memorie de proveniență:

````python
linie = tab[1, :]  # slicing -> view: aceleasi date ca tab

print("Demo indexare logica: tab=\n", tab)

valori = tab[~np.isnan(tab)]  # indexarea logica produce copie
valori[:] = 0
print("Dupa modificarea copiei: tab=\n", tab)  # tab NU s-a modificat

print(f"{np.shares_memory(tab, linie)=}\n{np.shares_memory(tab, valori)=}")
````

## Broadcasting

Broadcasting este un mecanism prin care se permite - în anumite circumstanțe - operarea cu matrice de dimensiuni ce nu sunt compatibile din punct de vedere al dimensiunii. De exemplu, urmând strict definiția matematică a adunării, matricele `a` și `b` de mai jos nu se pot aduna:

````python
a = np.array(
    [[0.0, 0.0, 0.0], [10.0, 10.0, 10.0], [20.0, 20.0, 20.0], [30.0, 30.0, 30.0]]
)
b = np.array([0.0, 1.0, 2.0])

print(f"a=\n{a}\n")
print(f"b=\n{b}\n")
````

Prin broadcasting, `b` se comportă ca și cum linia ar fi repetată pentru fiecare linie a lui `a`. NumPy nu copiază efectiv datele; repetarea este doar virtuală, ceea ce face ca operația să fie rapidă și economică în memorie.
<center>
    <img src="./images/broadcast1.png" width="600">
</center>

````python
# broadcasting
result = a + b
print(f"result=\n{result}\n")
````

Când se operează cu două tablouri, NumPy compară dimensiunile - atributul `shape` - element cu element, începând cu ultima dimensiune. Două dimensiuni sunt compatibile când:
1. sunt egale, sau
1. una din ele este 1

Dacă tablourile au număr diferit de dimensiuni, forma celui cu mai puține dimensiuni se completează cu 1 la stânga. De exemplu, `b` cu forma `(3,)` este tratat ca având forma `(1, 3)`, deci se poate aduna cu `a`, de formă `(4, 3)`.

Regulile de mai sus nu sunt îndeplinite, de exemplu, pentru:

````python
x = np.arange(4)
y = np.ones(5)
print(x.shape, y.shape)
# print(x+y) # ValueError: operands could not be broadcast together with shapes (4,) (5,)
````

## Vectorizare

Vectorizarea înseamnă exprimarea unei prelucrări asupra unor colecții de date ca operație pe tablouri întregi, în locul buclelor explicite scrise în Python care parcurg elementele unul câte unul. Bucla/bucclele se mută din interpretorul Python în cod compilat (C) din interiorul bibliotecii NumPy.

### Exemplu

Se dau două colecții de numere: prima conține distanțe parcurse, a doua timpul necesar pentru parcurgere. Se cere determinarea vitezelor corespunzătoare. Se va face implementare folosind ciclare (clasic) și vectorizare.

````python
distante = [10, 20, 23, 14, 33, 45]
timpi = [0.3, 0.44, 0.9, 1.2, 0.7, 1.1]
````

````python
# var 1: folosind ciclare
viteze = []
for i in range(len(distante)):
    viteze.append(distante[i] / timpi[i])

print("Viteze: ", viteze)
````

````python
# var 2: vectorizare
# vectorizarea numpy lucreaza peste tablouri, primul pas este obtinerea de tablouri din cele doua liste

distante_array = np.array(distante)
timpi_array = np.array(timpi)

# se folosesc operatii NumPy care trateaza tablourile in intregime. Codul C folosit pentru implementarea operatiilor NumPy
# foloseste facilitatile de executie Single Instruction Multiple Data (SIMD) din microprocesor.
viteze_array = distante_array / timpi_array

print(viteze_array)
````

### Beneficii

1. Execuție rapidă
1. Cod mai scurt și deseori mai clar

Exemplu: să se calculeze:
$$
\sum\limits_{i=0}^{N-1} (i\%3-1) \cdot i
$$

````python
# functie Python implementata naiv

N = 100000


def func_python(N):
    d = 0.0
    for i in range(N):
        d += (i % 3 - 1) * i
    return d


print(func_python(N))
````

````python
%timeit func_python(N)
````

````python
# functie rescrisa folosind facilitatile NumPy si vectorizare


def func_numpy(N):
    i_array = np.arange(N)
    return ((i_array % 3 - 1) * i_array).sum()


print(func_numpy(N))
````

````python
%timeit func_numpy(N)
````

Majoritatea operatorilor și a funcțiilor NumPy lucrează element cu element (sunt numite și universal functions, sau ufuncs) și într-un mod optimizat (SIMD). Următoarele sunt ufuncs:
- operatori aritmetici: `+ - * / // % **`
- operații pe biți: `& | ~ ^ >> <<`
- comparații: `< <= > >= == !=`
- funcții matematice: `np.sin`, `np.log`, `np.exp`, ...
- funcții speciale: `scipy.special.*`

Deși unele funcții din NumPy se regăsesc și în Python (ex: `sum`, `min`, `max`), folosirea de ufunc duce la execuție mai rapidă:

### Compilare just-in-time cu Numba

O parte considerabilă din codul Python poate beneficia de compilarea just-in-time (JIT) oferită de [Numba](https://numba.readthedocs.io/en/stable/). Pentru aceasta este nevoie ca funcția pentru care se vrea execuție rapidă să fie decorată cu `@jit`; ca urmare, funcția va fi tradusă în cod mașină, mult mai rapid decât codul interpretat. 

Exemplul de mai jos se folosește de aceeași implementare ca mai sus, doar cu folosirea de decorator. Pentru mai multe detalii recomandăm parcurgerea materialului [Numba](https://numba.readthedocs.io/en/stable/). Un decorator este o funcție care primește o altă funcție și întoarce o variantă modificată a ei; se aplică scriind `@nume_decorator` deasupra definiției. `@jit` întoarce o variantă compilată a funcției `func_python_numba`.

````python
from numba import jit

N = 100000

@jit
def func_python_numba(N):
    d = 0.0
    for i in range(N):
        d += (i%3-1) * i
    return d

%timeit func_python_numba(N)
````

Se observă timpul de execuție semnificativ mai redus față de variantele Python și cod vectorizat. Există însă limitări ale pachetului Numba, nu orice funcție beneficiază de compilarea JIT.

````python
@jit
def func_numpy_jit(N):
    i_array = np.arange(N)
    return ((i_array % 3 - 1 ) * i_array).sum()

%timeit func_numpy_jit(N)
````

### Exercițiu
Se dau $n$ puncte în spațiul bidimensional, prin coordonatele lor memorate în 2 vectori **x** și **y**. Să se determine care este cea mai apropiată pereche de puncte, considerând distanța euclidiană:
$$
d^2((x_i, y_i), (x_j, y_j)) = (x_i-x_j)^2 + (y_i-y_j)^2
$$

````python
n = 1000
x = rng.random(size=n)
y = rng.random(size=n)
````

Varianta 1: se calculează matricea pătratelor distanțelor de dimensiune n*n. d[i, j] va reprezenta pătratul distanței între punctul de coordonate (xi, yi) și cel de coordonate (xj, yj).

````python
%%timeit

d = np.empty((n, n))
for i in range(n):
    for j in range(n):
        d[i, j] = (x[i] - x[j])**2 + (y[i]-y[j])**2
````

````python
# calculul perechii de puncte i, j cu i != j pentru care distanta e minima
def pereche_apropiata(mat):
    # se face o copie a matricei pentru a nu modifica diagonala originalului
    mat = mat.copy()
    n = mat.shape[0]
    # distanta dintre un punct si el insusi este intotdeauna 0; se vor exclude aceste cazuri, setand infinit pentru diagonala principala
    i = np.arange(n)
    mat[i, i] = np.inf
    pos_flatten = np.argmin(mat)
    return pos_flatten // n, pos_flatten % n


# print(pereche_apropiata(d))
````

Varianta 2: broadcasting, vectorizare pentru calculul matricei d

`x[:, np.newaxis]` transformă vectorul de formă `(n,)` într-o matrice coloană, cu forma `(n, 1)`, iar `x[np.newaxis, :]` îl transformă într-o matrice linie, adică de forma `(1, n)`. Prin broadcasting, diferența lor are forma `(n, n)`: elementul `[i, j]` este `x[i] - x[j]`.

````python
%%timeit

dx = (x[:, np.newaxis] - x[np.newaxis, :]) ** 2
dy = (y[:, np.newaxis] - y[np.newaxis, :]) ** 2

d = dx + dy
````

Pentru afișarea perechii de puncte, comentați `%%timeit` din celula anterioară, rulați-o din nou, apoi rulați `print(pereche_apropiata(d))` într-o celulă de cod.

````python
# varianta cu numba

@jit
def compute_distances(x, y):
    n = x.shape[0]
    d = np.empty((n, n))
    for i in range(n):
        for j in range(n):
            d[i, j] = (x[i] - x[j]) ** 2 + (y[i] - y[j]) ** 2
    return d

%timeit d = compute_distances(x, y)
````
